```python
import numpy as np
import torch
import random

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
```

```python
labels = list(range(32))
```

```python
def split_labels(n_base, n_val, n_novel):
    assert n_base + n_val + n_novel == len(labels), "The sum of n_base, n_val, and n_novel must equal the total number of labels."

    np.random.shuffle(labels)
    base_labels = labels[:n_base]
    val_labels = labels[n_base:n_base + n_val]
    novel_labels = labels[n_base + n_val:]

    return base_labels, val_labels, novel_labels
```

```python
set_seed(42)
n_base = 16
n_val = 6
n_novel = 10

base_labels, val_labels, novel_labels = split_labels(n_base, n_val, n_novel)
print("Base labels:", base_labels)
print("Validation labels:", val_labels)
print("Novel labels:", novel_labels)
```
